You are provided with a dataset of SMS messages classified as spam and non-spam, ```spam.csv```.

**Goal:** Build a model that will classify a message as spam/non-spam using the Naive Bayes algorithm.

````python
import pandas as pd
from dataset import Dataset
from model import Model
````

````python
# Load dataset
df = pd.read_csv('spam.csv', encoding='latin-1')
df = df.iloc[:, [0, 1]] # Keep only label and message columns
df.columns = ['label', 'message']


print(f"Total Rows: {len(df)}")
print(df['label'].value_counts())
````

````python
# Prepare Data
X = df['message'].values
y = df['label'].values

# Initialize Dataset
ds = Dataset(X, y)
ds.split_dataset(val=0.1, test=0.1)

print(f"Train size: {len(ds.train[0])}")
print(f"Validation size:   {len(ds.val[0])}")
print(f"Test size:  {len(ds.test[0])}")

# Initialize and Train Model
model = Model(alpha=1.0)
model.fit(ds)
print("Model training completed.")
````

````python
# Validation Accuracy
val_acc = model.validation()
print(f"Validation Accuracy: {val_acc * 100:.2f}%")

# Test Accuracy
test_acc = model.test()
print(f"Test Accuracy:       {test_acc * 100:.2f}%")
````

````python
# Test your own messages
message = "URGENT! You have won a 1 week FREE membership in our £100,000 Prize Jackpot!"


prediction = model.inference(message)
print(f"Message: {message}...")
print(f"Prediction: {prediction.upper()}")
````
